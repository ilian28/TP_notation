Calcul des moyennes UE

```python
###Partie 1



n = int(input("Combien de notes ? : "))
total = 0
total_coeff = 0
for i in range(n):
    note = float(input("Entrer la note : "))
    coeff = float(input("Entrer le coefficient : "))
    total = total + note * coeff
    total_coeff = total_coeff + coeff
moyenne = total / total_coeff
print("La moyenne de l'UE est : ", moyenne)
```

```python
###Partie 2


total=0
total_coeff=0
note=[10,5,10,0]#Je crée une liste de toutes les notes
coeff=[10,0,18,12]#Je crée une liste avec touts les coeff attribuer à une note 

for i in range(0,4):#Pour toute entier i allant de 0 a 4   
    total=total+note[i]*coeff[i] 
    total_coeff=total_coeff+coeff[i]

moyenne=total/total_coeff
print("la moyenne de l'ue est de :",moyenne)
```
